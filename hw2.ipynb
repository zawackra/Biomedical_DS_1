```python
class MyData:

    def __init__(self, data):
        self.data = data
        self.mean = sum(data) / len(data)
        self.std = (sum((x - self.mean) ** 2 for x in data) / len(data)) ** 0.5


x = MyData([1, 2, 7, 5])

print(x.mean) 
print(x.std)  
```

```python

import numpy as np
a = np.array( [1, 6, 8, 4] ) 
integer=input("Pick an integer >0:")
def lag(a, integer):
    
    a_lag = np.empty(len(a), dtype=float)

    a_lag[:integer] = np.nan

    a_lag[integer:] = a[:-integer]

    return a_lag

```
